Q1: What are Artificial Neural Networks?
Artificial Neural Networks (ANNs) are computational models inspired by the structure and functioning of biological neural networks in the human brain. ANNs are composed of interconnected layers of nodes (neurons), and they process data by learning patterns and relationships. They are used in machine learning to solve complex problems, such as image recognition, natural language processing, and decision-making.

Q2: What are Layers in a Neural Network?
Layers in a neural network are the building blocks where data transformation occurs. They consist of a collection of neurons, each performing computations based on the inputs received. There are three main types of layers:

1. Input Layer: Receives the raw data.
2. Hidden Layers: Perform computations to learn complex patterns in the data.
3. Output Layer: Produces the final prediction or classification.


Q3: What are Neurons in a Neural Network?
Neurons are the basic computational units of a neural network. Each neuron receives input signals, processes them using a mathematical function, and transmits the output to the next layer. A neuron typically performs:

1. Weighted summation of inputs.
2. Application of an activation function to introduce non-linearity.
3. Passing the result as output.

Q4: What are Activation Functions in a Neural Network?
Activation functions in a neural network determine whether a neuron should be activated (fire) or not, introducing non-linear properties to the network. This enables the model to learn complex relationships. Common activation functions include:

1. Sigmoid: Maps inputs to a range between 0 and 1.
2. ReLU (Rectified Linear Unit): Outputs the input directly if it's positive, otherwise outputs 0.
3. Tanh: Maps inputs to a range between -1 and 1.
4. Softmax: Used in the output layer for multi-class classification problems.

Q5: What are the differences between a Biological Neuron and an Artificial Neuron?
A biological neuron is a natural cell in the brain and nervous system that transmits electrochemical signals to perform cognitive and physical functions. It consists of dendrites (receiving signals), a soma (cell body), and an axon (transmitting signals). Biological neurons communicate through synapses and adapt via synaptic plasticity.

An artificial neuron, on the other hand, is a mathematical model used in neural networks to simulate the behavior of a biological neuron. It processes numeric inputs by performing a weighted summation, adding a bias term, and applying an activation function to produce an output. Unlike biological neurons, artificial neurons are computational and learn through algorithms like gradient descent. Biological neurons are slower but highly complex, while artificial neurons are faster but simpler in design.
